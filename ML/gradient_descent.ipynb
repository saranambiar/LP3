```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
def f(x):
    return (x+3)**2

def gradient(x):
    return 2*(x+3)
```

```python
x = 2
learning_rate = 0.1
iterations = 50

path = [x]
```

```python
for i in range(iterations):
    slope = gradient(x)
    x = x - learning_rate * slope
    path.append(x)
    if i < 10:
        print("step", i + 1, "x =", round(x, 4), "y =", round(f(x), 4), "slope =", round(slope, 4))

print()
print("Final x: ", x)
print("Final y: ", f(x))
```

```python
xs = np.linspace(-6, 3, 200)           # 200 x values from -6 to 3
plt.plot(xs, f(xs), label="y = (x+3)^2")
plt.scatter(path, [f(p) for p in path], color="red", label="steps")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```
